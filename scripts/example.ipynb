## 3. Combine classifiers with majority vote

```python
from semantic_segmentation_benchmark.constants import SITN_MINKUNET, FLAI_REMAP, CLASSICAL_REMAP, SITN_FIRST_ITERATION_DL_REMAP
from semantic_segmentation_benchmark.metrics import ClassificationCombination
import os

dir_ref = r"D:\Raphael\Essais\1_TUILES_DE_REFERENCE"
dir1 = r"D:\Raphael\point-cloud-classifier\visualization\0_Predicted_Tiles_Flai"
dir2 = r"D:\Raphael\point-cloud-classifier\visualization\1_Predicted_Tiles_v1"
dir3 = r"D:\Raphael\point-cloud-classifier\visualization\2_Predicted_Tiles_Simon"
dir4 = r"D:\Raphael\point-cloud-classifier\visualization\4_Predicted_Tiles_Minkunet"
out_dir = r"D:\Raphael\point-cloud-classifier\visualization\10_combined_majority_vote"

for file in os.listdir(dir_ref):
    if file == "2545500_1198000.copc.laz":
        continue

    file1 = os.path.join(dir1, file)
    file2 = os.path.join(dir2, file)
    file3 = os.path.join(dir3, file)
    file4 = os.path.join(dir4, file)
    files = [file1, file2, file3, file4]

    remaps = [FLAI_REMAP, CLASSICAL_REMAP, SITN_FIRST_ITERATION_DL_REMAP, SITN_MINKUNET]
    output = os.path.join(out_dir, file)

    ClassificationCombination.combine_majority_vote(files, remaps, os.path.join(dir_ref, file), output, 2.5)
```
