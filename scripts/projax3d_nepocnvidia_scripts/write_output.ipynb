```python
from projax3d.io import Scene
import os

path = r"L:\raphael\projax-3d-models\data\Test_output_tiles"
output_path = r"L:\raphael\2_inference_output\data"

for scene_name in os.listdir(path):
    scene_test = Scene.from_parquet(os.path.join(path, scene_name), crs="EPSG:2056")
    scene_test.compute_footprint()
    scene_test.to_las(os.path.join(output_path, scene_name) + ".laz", columns={"classification": "minkunet_v3_prediction"}, extra_columns=["entropy"])
```
